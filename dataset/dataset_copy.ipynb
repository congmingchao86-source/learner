# 第一步：安装 conda-pack
- 打开你的 WSL Ubuntu 终端，执行：
conda activate ipad
- 然后安装打包工具：
conda install -c conda-forge conda-pack
# 第二步：打包你的 ipad Conda 环境
- 在 WSL Ubuntu 终端中执行：
conda -pack -n ipad -o /mnt/g/Udata/ipad.tar.gz
```
参数解释：
- -n ipad：指定要打包的 Conda 环境为 ipad。
- -o：指定打包文件的保存路径。
- /mnt/g/Udata/ipad.tar.gz：将环境压缩包保存到 Windows 的 G:\Udata 文件夹中。
打包完成后，你就能在 G 盘看到 ipad.tar.gz 文件。
```
运行后如果出现：
```python
(ipad) haha@DESKTOP-5D47ATB:~$ conda-pack -n ipad -o /mnt/g/Udata/ipad.tar.gz
Collecting packages...
CondaPackError: Cannot pack an environment with editable packages
installed (e.g. from `python setup.py develop` or
 `pip install -e`). Editable packages found:

- /home/haha/iPad-main
- /home/haha/iPad-main/Bench2DriveZoo
- /home/haha/iPad-main/nuplan-devkit
(ipad) haha@DESKTOP-5D47ATB:~$ 
```
说明：你的 ipad 环境中存在 3 个以可编辑模式（pip install -e）安装的 Python 项目，导致 conda-pack 拒绝打包。
# 第三步：忽略可编辑模式
解决方法:
```python
conda-pack -n ipad -o /mnt/g/Udata/ipad.tar.gz --ignore-editable-packages
```
我们可以使用 conda-pack 的 --ignore-editable-packages 参数，忽略这项检查，继续打包。\
*但要特别注意： 这种方式不会自动将上述三个项目的完整源码打包进去。因此，以后迁移环境时，还需要单独复制 /home/haha/iPad-main 项目目录。*
