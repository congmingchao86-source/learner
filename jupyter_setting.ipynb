## （5）升级版本：

````python
/* 全局基础 */
:root {
    --jp-content-font-family: "SimSun", "宋体", "Times New Roman", Times, serif !important;
    --jp-ui-font-family: "SimSun", "宋体", "Times New Roman", Times, serif !important;
}
.jp-RenderedHTMLCommon {
    font-family: "SimSun", "宋体", "Times New Roman", Times, serif !important;
}

/* ========== Markdown表格（单独：宋体14px，深色样式） ========== */
.jp-RenderedHTMLCommon table {
    border-collapse: collapse !important;
    width: 100% !important;
    font-family: "SimSun", "宋体", "Times New Roman", Times, serif !important;
    font-size: 14px !important;
}
.jp-RenderedHTMLCommon th {
    background-color: #2c3748 !important;
    color: #ffffff !important;
    border: 1px solid #4a5568 !important;
    padding: 10px 12px !important;
    text-align: center !important;
}
.jp-RenderedHTMLCommon td {
    background-color: #1a202c !important;
    color: #e2e8f0 !important;
    border: 1px solid #4a5568 !important;
    padding: 10px 12px !important;
    text-align: center !important;
}
/* .jp-RenderedHTMLCommon tr:nth-child(even) td {
    background-color: #222938 !important;
} */
.jp-RenderedHTMLCommon tr:hover td {
    background-color: #2d3748 !important;
}

/* ========== 标题：固定宋体，字号不改动 ========== */
.jp-RenderedMarkdown h1,
.jp-RenderedMarkdown h2,
.jp-RenderedMarkdown h3,
.jp-RenderedMarkdown h4,
.jp-RenderedMarkdown h5,
.jp-RenderedMarkdown h6 {
    font-family: "SimSun", "宋体", "Times New Roman", Times, serif !important;
}

/* ========== 正文 p、li：楷体，字号放大 22px【重点！这里删掉了td、th，不和表格冲突】 ========== */
.jp-RenderedMarkdown p,
.jp-RenderedMarkdown li {
    font-family: "KaiTi", "楷体", "SimSun", "宋体", "Times New Roman", Times, serif !important;
    font-size: 22px !important;
}

/* ========== 代码块保留等宽字体 ========== */
.jp-ThemedContainer .cm-editor {
    font-family: "JetBrains Mono", Consolas, monospace !important;
}
.jp-OutputArea-output pre {
    font-family: "JetBrains Mono", Consolas, monospace !important;
}

/* ========== 加粗文字（蓝紫渐变，**文字**自动生效） ========== */
.jp-RenderedHTMLCommon strong {
    background-image: linear-gradient(90deg, #4096ff, #722ed1);
    -webkit-background-clip: text;
    background-clip: text;
    color: transparent !important;
    -webkit-text-fill-color: transparent !important;
    font-weight: bold !important;
}

/* 单独调用渐变类 <span class="gradient-text">文字</span> */
.gradient-text {
  background-image: linear-gradient(90deg, #4096ff, #722ed1);
  -webkit-background-clip: text;
  background-clip: text;
  color: transparent;
  -webkit-text-fill-color: transparent;
}
````
