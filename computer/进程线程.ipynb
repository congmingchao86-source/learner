# 进程和线程

- 进程：**操作系统加载起来运行的程序实例**。模型代码只是这个进程里面的一部分业务逻辑。
一个进程里面可以不止一个模型，还包含数据接收、预处理、后处理、发布话题等逻辑。

感知进程（1个进程，独立内存空间）\
├─ 线程A：接收传感器数据（收点云/图像）\
├─ 线程B：数据预处理（去畸变、坐标转换）\
├─ 线程C：模型推理（调用深度学习模型）\
└─ 线程D：后处理、卡尔曼滤波、目标跟踪\
总结：
- 进程：**程序跑起来之后的独立容器**（装一堆逻辑、模型、变量）
- 线程：**容器里面一条条独立干活的流水线**，每条流水线负责一小块任务，可以多条流水线一起开工\
✍️线程**可以同时跑多个**（并发 / 并行）：
- CPU 多核：多个线程**真正同时并行执行**；
- CPU 单核：靠操作系统快速切换，宏观上看起来同时在跑（并发）。

1. **核显（集成显卡）**：GPU 直接封装在 CPU 里面，**没有独立的显卡板子**，这里只有 GPU，没有独立显卡。
2. **独立显卡**：就是我们常买的 N 卡 / A 卡，它是一块完整的卡，上面搭载 GPU。
3. 2. **显存和 GPU 的关系**

- GPU 负责**算数**（矩阵运算、卷积、渲染），它本身几乎不存数据；
- 显存负责**存放当前要算的数据**，GPU 需要什么，直接从显存读取，算完再放回显存；
- 两者搭配：GPU 算力再强，**显存不够，程序直接报错 OOM（Out of Memory）**，哪怕 GPU 算力富余也用不了。

举个你熟悉的例子：跑 BEVFormer
模型权重、图像、BEV 特征、中间特征图全部塞进显存。
如果显存太小，放不下，直接爆显存，程序崩溃。

3. 和核显的区别（承接刚才的问题）
核显**没有独立显存**，它是从系统内存里划一块出来当临时 “显存”，带宽低、速度慢，而且不叫 VRAM。
