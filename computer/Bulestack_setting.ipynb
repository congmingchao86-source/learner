# 模拟器网络配置


**在模拟器的浏览器中查看其IP地址：https://browserleaks.com/ip** \
**如果没有挂上梯子进行下面步骤：**

## 1.clash配置：


- 开启Clash Verge 的 TUN（虚拟网卡）模式（无法开启则需要安装：蓝色扳手图标  点击它）
- 开启全局模式
- 设置里面开启：区域网连接\
**可以在clash的连接界面查看，当模拟器使用网址刷新的时候，clash连接有没有反应**

## 2.windows检查

- 在cmd运行 ipconfig查看ip
- 查看电脑是个否有ADB：adb version

## 3.模拟器的ADB 调试

（1）点击蓝叠窗口右侧工具栏底部的齿轮图标，打开 BlueStacks 设置。\
（2）点击左侧的 「高级」。\
（3）找到 「Android 调试桥（ADB）」 或 「ADB 调试」。
这里是模拟器提供是地址

## 4.关键操作

- 电脑终端运行：dir "C:\Program Files\BlueStacks_nxt\HD-Adb.exe"\
显示 HD-Adb.exe：说明找到了蓝叠自带的 ADB。\
显示“找不到文件”：说明蓝叠可能安装在其他位置。\
**连接蓝叠模拟器**
- 如果存在HD-Adb.exe；运行(其中地址：127.0.0.1:5555是模拟器上给的)：\
"C:\Program Files\BlueStacks_nxt\HD-Adb.exe" connect 127.0.0.1:5555
| 部分 | 含义 |
| ---- | ---- |
| "C:\Program Files\BlueStacks_nxt\HD-Adb.exe" | 蓝叠自带的 ADB 程序 |
| connect | ADB 的连接命令 |
| 127.0.0.1:5555 | 要连接的地址和端口 |

(+)127.0.0.1 是一个特殊的 IP 地址，叫作 本机回环地址（Loopback Address）{<u>127.0.0.1 = Windows 电脑自己</u>}\
(+)5555 是 TCP 端口号。IP 地址用来确定连接哪台设备，端口号用来确定连接该设备上的哪个网络服务。
例如：\
127.0.0.1:5555\
    │       │\
    │       └── 端口号：5555\
    │\
    └────────── IP 地址：本机


**配置蓝叠使用 Clash Verge 代理**
- 最后运行："C:\Program Files\BlueStacks_nxt\HD-Adb.exe" -s 127.0.0.1:5555 shell settings put global http_proxy 10.3.0.136:7897
| 命令部分 | 含义 |
| ---- | ---- |
| HD-Adb.exe | 蓝叠自带的 ADB 工具 |
| -s | 指定要操作的安卓设备 |
| 127.0.0.1:5555 | 蓝叠的 ADB 连接地址 |
| shell | 在安卓系统中执行命令 |
| settings | 安卓系统设置管理工具 |
| put | 写入或修改一个设置值 |
| global | 安卓系统的全局设置数据库 |
| http_proxy | HTTP 代理配置项 |
| 10.3.0.136:7897 | 代理服务器 IP 和端口 |


**注意：**
- 其中10.3.0.136本机的ip地址（可以用过cmd输入ipconfig查看）
- 7897是clash的端口（在clash设置当中查看）
