## 1.logging用法：


```python
import logging
logger = logging.getLogger("__name__")
logger.info("Building Agent Input Loader")

logging.info("模型开始加载")
logging.warning("数据加载速度较慢")
logging.error("模型加载失败")
```

Python 的 logging 有不同的日志级别：

| 日志级别 | 用途 |
| ---- | ---- |
| DEBUG | 调试信息 |
| INFO | 程序正常运行的信息 |
| WARNING | 警告信息 |
| ERROR | 错误信息 |
| CRITICAL | 严重错误 |

**Python 在没有其他日志配置的情况下，默认只显示 WARNING 及以上级别的日志**，
因此：
```python
logger.info("Building Agent Input Loader")
```
