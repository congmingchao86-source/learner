# 一、Numpy数组和Pytorch Tensor最重要的三个区别
| 区别 | NumPy 数组 | PyTorch Tensor |
| ---- | ---- | ---- |
| CPU 计算 | ✅ | ✅ |
| GPU 加速 | 通常不直接支持 | ✅ |
| 自动求导 | 不支持 | ✅ |

````python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = x ** 2

y.backward()

print(x.grad)
````

PyTorch 自动计算出了：
$$
\frac{dy}{dx}=2x=4
$$

NumPy 本身没有这样的自动求导机制。

# 二、字典推导式：
{  
    键: 值  
    for 变量1, 变量2 in 可迭代对象  
}

````python
data = {
    "a": [1, 2, 3],
    "b": [4, 5]
}

result = {
    key: len(value)
    for key, value in data.items()
}
print(result)
````

# 三、\_\_post_init\_\_ 是什么？
它的作用是：在 @dataclass 自动生成的 \_\_init__() 执行完成后，自动执行 \_\_post_init__()。

例如：

````python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int

    def __post_init__(self):
        print("初始化完成！")

s = Student("Tom", 20)
````

# 四、@property 是 Python 中的一个装饰器，作用是：让类中的方法可以像普通属性一样访问，不需要加 ()

````python
#普通方法
class Student:
    def get_age(self):
        return 20

s = Student()

print(s.get_age())
````

````python
#采用装饰器
class Student:
    @property
    def age(self):
        return 20

s = Student()

print(s.age)
````

# 七、callable()
Python 内置函数，不用 import  
作用：判断一个对象能不能被调用（能不能加 () 执行），返回 True / False  
能 call：函数、类、方法 
不能 call：普通数字、字符串、tensor 实例、普通变量  

````python
def foo():
    pass

class Box:
    pass

x = 10
s = "hi"

print(callable(foo))   # True，函数可以调用
print(callable(Box))   # True，类可以调用（Box() 实例化）
print(callable(x))     # False，数字不能加()
print(callable(s))     # False，字符串不能加()
````

# 1、$ getattr() $
Python 内置函数，不用 import  
作用：获取对象的属性 / 方法，等价于 obj.xxx，但是名字可以用字符串传入  
$ getattr(object, name, default=None) $   
- object：目标对象  
- name：属性名，必须是字符串  
- default（可选）：找不到属性时返回这个值；不写的话找不到会抛 AttributeError

````python
import torch
x = torch.tensor(2.0, requires_grad=True)
y = x**2
# 等价于 x.grad
print(getattr(x, "grad"))
print(x)
print(y)

# 等价于 x.backward
f = getattr(x, "backward")
f()   # 等同于 x.backward()
print(f())
````

````python
from dataclasses import dataclass
@dataclass
class Box:
    x: float
    y: float

b = Box(1.0,2.0)
print(getattr(b, "x")) # 1.0
````

# 2、$ slice() $
Python 内置函数，不用 import  
作用：创建切片对象，等价于$ a[start:end:step] $里的 start:end:step，把切片规则存成变量，复用。

````python
lst = [0,1,2,3,4,5,6,7]

# 写法1：直接切片
print(lst[1:5])   # [1,2,3,4]

# 写法2：先用slice造切片对象
s = slice(1,5)
print(lst[s])     # [1,2,3,4]
print(s)

s = slice(1,7,2)
print(lst[s]) # [1,3,5]
````

- n：输出匹配行的行号
- "split: test\|split: mini\|navsim_log_path:"：搜索正则
- \\| 是正则里的 或
- 匹配三种子模式任意一个：
- split: test
- split: mini
- navsim_log_path:
- 最后路径：搜索根目录，NavSim Hydra 配置文件夹

# 8.yaml配置文件中的常用函数$oc.env$：

```python
cache_path: ${oc.env:NAVSIM_EXP_ROOT}/train_metric_cache
```
- oc.env 是 OmegaConf 内置 Resolver（解析器），作用：**读取系统环境变量**
- 读环境变量 NAVSIM_EXP_ROOT 的值
- 拼接后缀 /train_metric_cache，得到最终缓存目录

等价于python的视角：
```python
import os
cache_path = os.getenv("NAVSIM_EXP_ROOT") + "/train_metric_cache"
```
进阶语法：**给默认值（非常常用）:逗号后面写默认值，环境变量不存在时 fallback：**
```python
# 如果 NAVSIM_EXP_ROOT 没设置，默认用 /data/navsim_exp
cache_path: ${oc.env:NAVSIM_EXP_ROOT,/data/navsim_exp}/train_metric_cache
```
